````python
# пример применения метода ко всем элементам списка

list(map(str.lower, ['Я БОЛЬШЕ', 'КОГО ТЫ БОЛЬШЕ']))
````

````python
def input_fio_list():
    """
    Функция для ввода ФИО до слова "Стоп"
    Возвращает список ФИО
    """
    fio_list = []
    while True:
        s=input().strip()
        if s.lower()=="стоп": break
        if s: fio_list.append(s)
    return fio_list


def capitalize_fio(a):
    """
    Функция для преобразования регистра ФИО
    Принимает список ФИО, возвращает список с преобразованным регистром
    """
    # Ваш код здесь для подпункта b)
    return list(map(lambda x: x.title(), a))


def format_fio(a):
    r=[]
    for s in a:
        p=s.split()[:3]
        if len(p)<2:
            r.append(p[0]); continue
        t=p[0]+" "+p[1][0]+"."
        if len(p)>2: t+=" "+p[2][0]+"."
        r.append(t)
    return ", ".join(r)


# Ввод данных
fio_list = input_fio_list()

# Преобразование регистра
capitalized_fio = capitalize_fio(fio_list)

# Форматирование ФИО
result = format_fio(capitalized_fio)

# Вывод результата
print(result)
````

````python
import time
import random
from functools import wraps


def timer(arg_func):
    """
    Декоратор, который измеряет время выполнения функции
    и подсчитывает количество её запусков
    """
    # Ваш код здесь для подпункта a) и e)

    import time
    total_time = 0  # прошедшее время всех вызовов всех функций с декоратором
    counts = {}      # счетчик вызовов декоратора

    @wraps(arg_func)
    def wrapper(*args, **kwargs):
        nonlocal total_time

        start = time.time() # вызываем из библиотеки time функцию time для начала отсчета
        result = arg_func(*args, **kwargs) # вызываем пользовательскую функцию
        time_taken = time.time() - start # сколько прошло времени со старта функции
        print(f'Taken time {time_taken} for function {arg_func.__name__}')
        total_time += time_taken
        return result

    wrapper.counts = counts
    wrapper.total_time = total_time

    return wrapper


@timer
def square_sum(left, right):
    """
    Функция вычисляет сумму квадратов чисел в диапазоне [left, right]
    """
    # Ваш код здесь для подпункта b)
    return sum(i ** 2 for i in range(int(left), int(right) + 1))



@timer
def take_greatest_even(n, k):
    """
    Функция генерирует список из n случайных чисел,
    сортирует по убыванию и возвращает k наибольших четных чисел
    """
    # Ваш код здесь для
    arr = [random.randint(0, 100) for _ in range(int(n))]
    arr.sort(reverse=True)
    return [x for x in arr if x % 2 == 0][:int(k)]



# Ввод данных для square_sum
range_input = input().split()
left, right = map(int, range_input)

# Ввод данных для take_greatest_even
array_input = input().split()
n, k = map(int, array_input)

res1 = square_sum(left, right)
res2 = take_greatest_even(n, k)
print(res1)
print(res2)


````

````python
class Animal:
    """
    Базовый класс для всех животных
    """
    def __init__(self, name, animal_type, age, sound):
        self.name = name
        self.animal_type = animal_type
        self.age = int(age)
        self.sound = sound

    def make_sound(self):
        print(f"{self.name} says: {self.sound}")
        pass

    def info(self):
        print(f"name: {self.name}, type: {self.animal_type}, age: {self.age}")



class Mammal(Animal):
    """
    Класс Млекопитающие, наследуется от Animal
    """
    def __init__(self, name, animal_type, age, sound, has_fur):
        super().__init__(name, animal_type, age, sound)
        self.has_fur = bool(has_fur)
        self.sleeping = False

    def feed_young(self):
        print(f"{self.name} feed milkk")

    def _sleep(self):
        if not self.sleeping:
            self.sleeping = True
            print(f"{self.name} наелся и спит")
        else:
            self.sleeping = False
            print(f"{self.name} проснулся")


    def info(self):
        print(f"name: {self.name}, type: {self.animal_type}, age: {self.age}, sleeping: {self.sleeping}")



class Bird(Animal):
    """
    Класс Птицы, наследуется от Animal
    """
    def __init__(self, name, animal_type, age, sound, wingspan):
        super().__init__(name, animal_type, age, sound)
        self.wingspan = float(wingspan)

    def fly(self):
        print(f"{self.name} летает")


    @staticmethod
    def drozdov_words():
        print("круточе пение птиц")



class Lion(Mammal):
    """
    Класс Лев, наследуется от Mammal
    """
    def make_sound(self):
        print("рычит...")

    def feed_young(self):
        print("лев кормит детей")


class Dog(Mammal):
    """
    Класс Собака, наследуется от Mammal
    """
    def make_sound(self):
        print("гав гав")

    def feed_young(self):
        print("собака кормит дете")



class Eagle(Bird):
    """
    Класс Орел, наследуется от Bird
    """
    def make_sound(self):
        print("🇺🇸🦅")

    def fly(self):
        print(f"🇺🇸🦅 прыгнул высоко {self.wingspan}m")



class Nightingale(Bird):
    """
    Класс Соловей, наследуется от Bird
    """
    def make_sound(self):
        print("соловей поет...")

    def fly(self):
        print("летает между деревьями")



lion = Lion("simba", "lion", 5, "rrr", True)
dog = Dog("buddy", "dog", 3, "woof", True)
eagle = Eagle("aquila", "eagle", 4, "screech", 2.1)
nightingale = Nightingale("phil", "nightingale", 2, "trill", 0.25)

for a in (lion, dog, eagle, nightingale):
    a.make_sound()

lion.feed_young()
dog.feed_young()
eagle.fly()
nightingale.fly()
lion._sleep(); lion._sleep()

flying_animals = [x for x in (lion, dog, eagle, nightingale) if isinstance(x, Bird)]
feeding_animals = [x for x in (lion, dog, eagle, nightingale) if isinstance(x, Mammal)]

print("\nЛетающие животные:", [x.name for x in flying_animals])
print("Кормящие детенышей:", [x.name for x in feeding_animals])
````

````python
def create_cipher_dict(n):
    """
    Создает словарь для шифра Цезаря со сдвигом n
    Возвращает словарь, где ключ - исходная буква, значение - зашифрованная
    """
    assert -25 <= n <= 25, "Сдвиг должен быть от -25 до 25"
    cipher_dict = {}

    for i in range(26):
        a = chr(ord('A') + i)
        b = chr(ord('A') + (i + n) % 26)
        cipher_dict[a] = b
    for i in range(26):
        a = chr(ord('a') + i)
        b = chr(ord('a') + (i + n) % 26)
        cipher_dict[a] = b
    return cipher_dict


def encode(text, cipher_dict):
    """
    Шифрует текст using cipher_dict
    """
    result = []
    # Ваш код здесь для подпункта b)
    # Пройдите по каждому символу текста и замените его согласно словарю
    # Нешифруемые символы оставьте как есть
    for ch in text:
      result.append(cipher_dict.get(ch, ch))
    return ''.join(result)


def decode(text, cipher_dict):
    """
    Дешифрует текст using cipher_dict
    """
    # Создаем обратный словарь для дешифрования
    reverse_dict = {v: k for k, v in cipher_dict.items()}
    result = []
    # Ваш код здесь для подпункта c)
    # Пройдите по каждому символу текста и замените его согласно обратному словарю
    for ch in text:
        result.append(reverse_dict.get(ch, ch))
    return ''.join(result)


def process_messages():
    """
    Основная функция для обработки сообщений
    Принимает ввод до слова "СТОП" и записывает результаты в файл
    """
    # Получаем сдвиг от пользователя
    n = int(input().strip())

    # Создаем словарь для шифрования
    cipher_dict = create_cipher_dict(n)

    # Открываем файл для записи результатов
    with open("results.txt", "w", encoding="utf-8") as f:

        # Цикл обработки сообщений до слова "СТОП"
        while True:
            input_text = input().rstrip("\n")
            if input_text.lower() == "стоп":
                break
            command = input().strip()
            if command.lower() == "стоп":
                break
            if command == "decode":
                output_text = decode(input_text, cipher_dict)
            elif command == "encode":
                output_text = encode(input_text, cipher_dict)
            else:
                continue
            f.write(f"сдвиг: {n}; ввод: {input_text}; вывод: {output_text};\n")
            nxt = input().strip()
            if nxt.lower() == "стоп":
                break
            n = int(nxt)
            cipher_dict = create_cipher_dict(n)



# Запуск основной функции
process_messages()
print("Результаты сохранены в файл results.txt")
````

````python
import random

def find_artefact(items):
    """
    Использует filter и lambda для поиска предметов, содержащих слово 'камень'
    Возвращает отфильтрованный список
    """
    # Ваш код здесь для подпункта a)
    stone_items = list(filter(lambda x: "камень" in str(x).lower(), items))

    # Используйте filter и lambda функцию
    return stone_items


def check_artefact(items):
    """
    Использует map и lambda для проверки, является ли предмет из списка Камнем Желаний
    Возвращает список True/False
    """

    is_wish_stone = list(map(lambda x: str(x).strip().lower() == "камень желаний", items))

    return is_wish_stone


def recover(people, artefacts):
    """
    Восстанавливает подданных, если найден Камень Желаний
    Возвращает список имен восстановленных подданных или исходный список
    """
    # Разделяем людей на имена и животных


    # Проверяем, найден ли Камень Желаний

    has_ws = any(map(lambda x: str(x).strip().lower() == "камень желаний", artefacts))
    if has_ws:
        recovered = list(map(lambda s: s.split(",")[0].strip(), people))
    else:
        recovered = list(map(lambda s: s, people))
    # Ваш код здесь для подпункта c)
    # Используйте map и lambda функцию

    return recovered


def save_the_kingdom():
    """
    Основная функция спасения королевства
    Объединяет все предыдущие функции и записывает результаты в файл
    """
    # Ввод предметов до первого "Стоп"
    items = []
    print("Введите предметы, которые могут встретиться в лесу (для завершения введите 'Стоп'):")
    while True:
        s = input().strip()
        if s.lower() == "стоп": break
        if s: items.append(s)


    # Ввод подданных до второго "Стоп"
    people = []
    print("\nВведите подданных в формате 'имя, животное' (для завершения введите 'Стоп'):")

    while True:
        s = input().strip()
        if s.lower() == "стоп": break
        if s: people.append(s)


    # Случайный выбор 5 предметов из леса (с повторами)
    found_items = random.choices(items, k=5) if items else []
    stone_items = find_artefact(found_items)

    # Проверка, есть ли среди них Камень Желаний
    stone_check = check_artefact(stone_items)

    # Восстановление подданных
    result_people = recover(people, stone_items)

    # Ваш код здесь для подпункта d)
    # Запись результатов в файл
    with open("kingdom_results.txt", "w", encoding="utf-8") as f:
        f.write(f"Найдены предметы: {found_items}; Камень желаний: {any(stone_check)}; Спасенные: {result_people};\n")

    # Вывод результата
    if any(stone_check):
        print("\nКоролевство спасено! Все подданные восстановлены!")
    else:
        print("\nКамень Желаний не найден. Подданные остаются в облике животных.")

    print(f"Результаты сохранены в файл kingdom_results.txt")


# Запуск основной функции
save_the_kingdom()
````
